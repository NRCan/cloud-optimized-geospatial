### Q: Can you show us an example of downloading and loading data from your dataset? / Q : Pouvez-vous nous montrer un exemple de téléchargement et de chargement de données à partir de votre jeu de données ?

*EN* : Examples of how to discover and access the dataset are available under the NRCan [Cloud Optimized Geospatial Data Access github repository](https://nrcan.github.io/cloud-optimized-geospatial/).

Below is an example of pystac-client to list all items from the dataset.

*FR* : Des examples de découverte et d'accès aux données sont présentées dans le dépôt GitHub de Ressources naturelles Canada (RNCan) consacré à l’accès aux données géospatiales optimisées pour le nuage : https://nrcan.github.io/cloud-optimized-geospatial/

Un exemple utilisant la librairie pystac-client pour faire la liste des items de la collection est présenté ci-dessous


```python
# Browse available items from the banff collection 
import pystac_client

# Link to ccmeo datacube stac-api
stac_root = "https://datacube.services.geo.ca/stac/api"
# Initialize the STAC client
catalog = pystac_client.Client.open(stac_root)

# Define search parameters
# In this example, we filter on the mosaic collection of HRDEM tiles
collections = ['banff']

# Build the search query with the filtering parameters
search = catalog.search(
    collections=collections
    ) 

# Use the pagination to improve efficiency
links=[]
# Iterate over each returned page and get the COG links
for page in search.pages():
    for item in page:
        for asset_name, asset in item.assets.items():
             # Only get links for asset role == 'data'
             if "data" in asset.roles:
                 links.append(asset.href)
```

*EN* : To read the header or a portion of a COG from the bucket, refer to the available examples here : https://nrcan.github.io/cloud-optimized-geospatial/example-cogs/

*FR* : Pour lire l’en-tête ou une section d’un fichier COG stocké dans le compartiment, reportez-vous aux exemples disponible ici : https://nrcan.github.io/cloud-optimized-geospatial/example-cogs/
